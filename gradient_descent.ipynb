```python
import numpy as np

def J(x):
    return x*2 + 2*x + 1

def dJ(x):
    return 2*x + 2

def gradient_descent(alpha, x0, max_iter=1000):
    x = x0
    for _ in range(max_iter):
        grad = dJ(x)
        x = x - alpha * grad
    return x

# Ejemplo de uso

alpha = 0.1
x0 = 5.0
x_min = gradient_descent(alpha, x0)
print(f"Mínimo encontrado en x = {x_min:.4f}")
```
